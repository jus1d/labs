````python
def properties(tag: str, xs: npt.NDArray[Any]):
    print(f"Массив: {tag}")
    print(f"  shape: {xs.shape}, dtype: {xs.dtype}, ndim: {xs.ndim}, size: {xs.size}")
    print(f"  Элементы:\n{xs}\n")
````

````python
array_1d = np.array([1, 2, 3, 4, 5])
properties('1d', array_1d)

array_3x3 = np.arange(1, 10).reshape(3, 3)
properties('3x3', array_3x3)

zeroes = np.zeros((5, 5))
properties('zeroes', zeroes)

ones = np.ones((3, 4))
properties('ones', ones)

eye4 = np.eye(4)
properties('eye', eye4)
````
